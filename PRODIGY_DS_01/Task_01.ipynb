# Task 01 - Population Distribution Visualization
Create a histogram to visualize the distribution of population across countries and regions.
The dataset used for this task is the World Bank population dataset provided by Prodigy InfoTech.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import os

print(os.getcwd())
```

```python
os.listdir()
```

```python
data = pd.read_csv("population.csv", skiprows=4)

data.head()
```

```python
data.shape
data.columns
data.info()
```

```python
population_2024 = pd.to_numeric(data["2024"], errors="coerce")
population_2024 = population_2024.dropna()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(population_2024, bins=30, edgecolor="black")

plt.title("Distribution of Population Across Countries and Regions in 2024")
plt.xlabel("Population")
plt.ylabel("Number of Countries/Regions")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(population_2024, bins=30, edgecolor="black")
plt.xscale("log")

plt.title("Distribution of Population Across Countries and Regions in 2024")
plt.xlabel("Population (log scale)")
plt.ylabel("Number of Countries/Regions")

plt.show()
```

## Conclusion

The histogram shows that population is not evenly distributed across countries and regions. A large number of countries have relatively smaller populations, while only a few countries have extremely large populations. The distribution is therefore strongly right-skewed.
